```haskell
-- Version 1
import Numeric (showFFloat)
 
-- Round a number to n significant figures
toSigFigs :: Int -> Double -> Double
toSigFigs n x
  | x == 0    = 0
  | otherwise = let d = ceiling (logBase 10 (abs x)) :: Int
                    factor = 10 ^ (n - d)
                in fromIntegral (round (x * factor) :: Int) / factor
```

```haskell
main = do
  let sqrtPart  = sqrt 45.0
      mulPart   = 3.2 * 4.8
      result    = sqrtPart + mulPart
      rounded   = toSigFigs 1 result
 
  putStrLn "=== Estimating sqrt(45) + 3.2 * 4.8 ==="
  putStrLn ""
  putStrLn $ "  sqrt(45)      = " ++ showFFloat (Just 6) sqrtPart ""
  putStrLn $ "  3.2 * 4.8     = " ++ showFFloat (Just 6) mulPart  ""
  putStrLn $ "  Full result   = " ++ showFFloat (Just 6) result   ""
  putStrLn ""
  putStrLn $ "  To 1 sig fig  = " ++ show rounded
```

```haskell
-- Version 2
-- Round a number to n significant figures
toSigFigs :: Int -> Double -> Double
toSigFigs n x
  | x == 0    = 0
  | otherwise = let d      = ceiling (logBase 10 (abs x)) :: Int
                    shift  = fromIntegral (n - d) :: Double
                    factor = 10 ** shift          -- (**) handles negative exponents
                in fromIntegral (round (x * factor) :: Int) / factor
```
