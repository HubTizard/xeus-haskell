## FL Exercise 3.3 foldr and foldl ##

__foldr__

![](images/foldr.png)

```haskell
foldr (:) [] "123"
```

foldr f e (1:2:3:[]) = 1 `f` (2 `f` (3 `f` e))

You can imagine foldr as some state-machine-evaluator, too:

f is the transition,

f :: input -> state -> state
and e is the start state.

e :: state
foldr (foldRIGHT) runs the state-machine with the transition f and the start state e (nil) over the list of inputs, starting at the right end. Imagine f in infix notation as the pacman coming from-RIGHT.

To make this clear, imagine the function (-) as transition:

- foldr (-) 100 [1]         = -99 = (1-(100))
- foldr (-) 100 [2,1]       = 101 = (2-(-99)) = (2-(1-(100)))
- foldr (-) 100 [3,2,1]     = -98 = (3-(101))
- foldr (-) 100 [4,3,2,1]   = 102 = (4-(-98))
- foldr (-) 100 [5,4,3,2,1] = -97 = (5-(102))

foldl (foldLEFT) does the same from-LEFT, but the transition function, written in infix notation, takes its input argument from right. So the machine consumes the list starting at the left end. Pacman consumes the list from-LEFT with an open mouth to the right, because of the mouth (b->a->b) instead of (a->b->b).

https://stackoverflow.com/questions/3950508/foldr-and-foldl-further-explanations-and-examples

```haskell
foldr (\y ys -> ys ++ [y]) [] [1,2,3]
```

```haskell
f y ys = ys ++ [y]
```

__foldl__

![](images/foldl.png)

foldl (foldLEFT) does the same from-LEFT, but the transition function, written in infix notation, takes its input argument from right. So the machine consumes the list starting at the left end. Pacman consumes the list from-LEFT with an open mouth to the right, because of the mouth (b->a->b) instead of (a->b->b).

- foldl (-) 100 [1]         = 99 = ((100)-1)
- foldl (-) 100 [1,2]       = 97 = (( 99)-2) = (((100)-1)-2)
- foldl (-) 100 [1,2,3]     = 94 = (( 97)-3)
- foldl (-) 100 [1,2,3,4]   = 90 = (( 94)-4)
- foldl (-) 100 [1,2,3,4,5] = 85 = (( 90)-5)

```haskell
foldl (-) 100 [1,2,3,4,5]
```

```haskell
-- foldl and scanl require an explicit start value (or seed)
-- compare with fold1 and scanl1 which take the first element as the seed to generate the triangle numbers
foldl1 (+) [1, 2, 3, 4, 5]  -- gives just 15
scanl1 (+) [1, 2, 3, 4, 5]  -- gives [1, 3, 6, 10, 15]
```

Exercise Handshake problem
The handshake problem: n people each shake hands with every other person exactly once — how many handshakes in total?

Solution The maths:

Each of n people could shake hands with n−1 others, giving n(n−1) — but that counts every handshake twice (once for each participant), so divide by 2:

$T(n) = \frac{n(n-1)}{2}$

Notice this is the triangle number T(n−1) — so the handshake problem and the triangle number problem are the same thing, shifted by one. The sequence of handshakes 0, 1, 3, 6, 10, 15... is just the triangle numbers T(0), T(1), T(2)...

```haskell
-- Formula: each of n people shakes hands with (n-1) others,
-- divide by 2 to avoid counting each handshake twice.
handshakes :: Integer -> Integer
handshakes n = n * (n - 1) `div` 2
 
-- Equivalently, handshakes(n) = triangle number T(n-1)
-- e.g. 5 people -> T(4) = 10 handshakes
 
-- Infinite list of (people, handshakes) pairs
handshakeTable :: [(Integer, Integer)]
handshakeTable = [(n, handshakes n) | n <- [1..]]
```

```haskell
main = do
  putStrLn "Handshake Problem: n * (n-1) / 2"
  putStrLn ""
  putStrLn "  People  Handshakes"
  putStrLn "  ------  ----------"
  mapM_ printRow (take 20 handshakeTable)
  where
    printRow (n, h) =
      putStrLn $ "  " ++ pad 6 n ++ "  " ++ show h
 
    pad width n =
      let s = show n
      in  replicate (width - length s) ' ' ++ s
 
```

```haskell
main
```
